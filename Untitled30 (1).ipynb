```python
my_dict = {"name": "zaha", "age": 18, "city": "lahore","eye":'blue', "hair":'black', "country": "pakistan", "skin ": "pale", "residence": "wapda town" }
 

# 2.get(key[, default]): Returns the value for key if key is in the dictionary, else default.

age = my_dict.get("age")
print(f"Age: {age}")
country = my_dict.get("country", "Not found")
print(f"Country: {country}")
```
